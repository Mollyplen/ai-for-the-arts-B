#### Initial Engagement with the Object

This initial appeal of this object came from the difference I perceived between it and other formal historical records. It felt more personal and layered, with a variety of different materials to explore. The journal records and details the routes sailed, weather conditions, and practical aspects of sailing, however, also offers a more human perspective, with insight into what daily life was like for those on board, including their relationships, daily activities, and social environments./n
