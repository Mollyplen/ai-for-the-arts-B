# MS Murray 167: From Archive to Interactive Experience

## Introduction & Context

This project explores the object MS Murray 167: Notes of Cruises in the “Raven” and “Wave” (1850–1866), held in the Archives and Special Collections at the University of Glasgow. The journal, compiled by Andrew Macgeorge, documents over ten years of sailing journeys along the west coast of Scotland and contains handwritten entries, sketches, diagrams, photographs, newspaper clippings, and reflections linked to James Smith of Jordanhill. The text is a detailed record of their travels, social life, and personal experiences during the mid-nineteenth century.
