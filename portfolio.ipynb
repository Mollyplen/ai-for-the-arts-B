# MS Murray 167: From Archive to Interactive Experience

## Introduction & Context

This project explores the object MS Murray 167: Notes of Cruises in the “Raven” and “Wave” (1850–1866), held in the Archives and Special Collections at the University of Glasgow. The journal, compiled by Andrew Macgeorge, documents over ten years of sailing journeys along the west coast of Scotland and contains handwritten entries, sketches, diagrams, photographs, newspaper clippings, and reflections linked to James Smith of Jordanhill. The text is a detailed record of their travels, social life, and personal experiences during the mid-nineteenth century.

#### Initial Engagement with the Object

This initial appeal of this object came from the difference I perceived between it and other formal historical records. It felt more personal and layered, with a variety of different materials to explore. The journal records and details the routes sailed, weather conditions, and practical aspects of sailing, however, also offers a more human perspective, with insight into what daily life was like for those on board, including their relationships, daily activities, and social environments.

### The Challenge of Accessibility

However, even with so much to learn from these pages, they are extremely difficult to read for modern audiences. The handwriting difficult for an untrained eye to transcribe, and it would be almost impossible to view or study the journal in a public setting, such as a museum. The entries vary in detail, and despite being written in English, the older style of language and sentence structure can be difficult to follow, making the historic value of the journal inaccessible to those unfamiliar with archival materials.

### From Archive to Digital Experience

This project responds to that challenge by transforming the journal into an interactive digital experience, beyond a simple digitisation of the material, which allows interested parties to still navigate it in their own way. It is not designed as a rigid, linear experience, but instead offers multiple points of interaction. The journal is presented as a digital book, where users move through sections using a page-based format. Key passages are curated into edited extracts, supported by additional layers that highlight human detail and encourage critical reflection.

### User Interaction and Engagement

*Figure 4: Digital interface of MS Murray 167, showing the journal reimagined as an interactive book*

Features such as themed sections, image-led pages, and the option to view both full and highlighted transcriptions, allow users to choose how deeply they engage with the material. This allows them to focus on what interests them most, whether visual elements, key extracts, or the full archival text. As a result, the level of detail gained from the journal is flexible and shaped by their own curiosity, creating a more personal and engaging experience.

### Interpretation and Perspective

The aim is not only to improve access, but to actively reshape how the journal is experienced. Rather than presenting the text as a static historical document, it foregrounds selective interpretation, guiding users towards key moments that reveal emotion, relationships, and social context. By structuring the material into curated extracts, layered interactions, and optional depth, users are encouraged to move beyond passive reading and question what is being shown, what is emphasised, and what may be missing. In this way, the journal becomes not just something to read, but something to explore, interpret, and consider critically.

*Figure 5: Interface showing the use of curated extracts and interpretive layers in the digital journal* 

A central focus of the project is perspective. The journal is approached as a subjective account shaped by the author’s experiences and social position, as opposed to a basic neutral record. This becomes particularly significant in the sections dealing with illness, memory and loss, as well as in the contrast between private journal entries and public accounts found in the newspaper clippings at the end of the journal, and how they are shaped by the perspectives of relatively wealthy individuals, influencing and potentially narrowing our understanding of this period.

### Overall Approach

Overall, this project offers a new way of engaging with MS Murray 167, combining archival material with digital design and interpretive framing. It demonstrates how historical documents can be re-presented in ways that remain faithful to the original, while making them more accessible, interactive, and open to critical exploration.

## Project Objectives & Audience

### Audience

The design of the project caters to a broad, non-specialist audience, including members of the general public and younger users who may encounter the journal in a museum or digital heritage setting, prioritising clarity, usability and engagement. The structure and interface use a simple design, similar to other interactive museum displays or digital exhibitions.


*Figure 6: Example of the clear and accessible interface designed for a broad, non-specialist audience*

### Objectives

A key objective was to make the journal readable in a practical sense by breaking down long handwritten entries into key ideas and curated extracts, while still preserving access to the full transcription. However, readability alone was not the primary goal. A further aim was to highlight the emotional and human elements within the journal, drawing attention to moments of humour, illness, routine and loss that might otherwise be overlooked when reading the text in full. By focusing on these aspects, the project encourages a wider audience to connect more meaningfully with the material.

Alongside this, it is intentionally designed to encourage active engagement through prompts for reflection. Visitors are invited to consider how the narrative is shaped, whose perspective is being represented, and what the entries reveal about the broader social and historical context. The combined structure, interaction, and curation provides opportunities for users to make meaning from the journal as they move through it, shaping their own experience. 


## Methodology

AI tools were used throughout the project to support both the design process and the presentation of the journal. Figma AI was used to assist with layout generation and prototyping, helping to establish consistent page structures and refine the organisation of content. This allowed for quicker experimentation with different layouts and improved the overall efficiency of the design process.

*Figure 6: Figma AI-generated layout prototype used to structure early page designs*

Text-based AI tools were used to support the structuring and refinement of written content, particularly when working with archival language. These tools helped clarify phrasing and organise extracts into more readable formats, while still maintaining the original meaning of the text.

*Figure 7: Example of AI-assisted text refinement for archival language*

Image-based AI tools were used in two ways - for cropping and cleaning photographed images to improve their clarity in a digital format and adding colour to selected images using tools such as Gemini. These models are trained on large datasets of colour images and learn to predict plausible colour values for grayscale inputs. Whilst the results are not historically accurate, they help engage users more deeply. The unexpected coloured images offer an interpretive layer that encourages imagination of what the scenes may have looked like in real life.

*Figure 8: Original archival image and AI-colourised version*

The initial plan was to use Transkribus, an AI-powered handwritten text recognition tool, to automate transcription. However, despite testing multiple publicly available models, the outputs were largely hard to make sense of and unusable. This exposed a key limitation of AI in handling complex historical handwriting and reinforced the importance of human interpretation early in the project, particularly in recognising letterforms and contextual meaning that the model failed to capture.

*Figure 9: Example of failed AI transcription output using Transkribus*

In general, these tools operate through a similar machine learning workflow. The input consists of prompts, images, or text provided by the user. The model, trained on large datasets, identifies patterns within the input and generates an output based on learned relationships. These outputs are probabilistic, meaning they are based on likelihood rather than fixed rules, which is why human review remains essential.

*Figure 10: Simplified AI workflow used in the project*

A key part of this project was clearly defining the difference between human and AI roles. The process was strongly human led, with AI used as a supporting tool rather than a decision maker. The use of AI is transparent within the design, with labels such as “archival”, “edited extract”, and “AI-assisted” distinguishing between original material, interpreted content, and AI-supported elements.

*Figure 11: Example of content labels distinguishing archival, edited, and AI-assisted material*

I remained the decision-maker throughout the process, selecting content, determining how it should be structured, and shaping audience experience of the journal. This involved identifying key extracts, interpreting their meaning, and deciding which aspects would be most engaging or significant. I also maintained a quality control role when working with AI outputs, particularly with transcription. AI-generated text was checked carefully against the original manuscript to correct inaccuracies and “hallucinated” paragraphs, ensuring the final content remained as close to the original as possible.

*Figure 12: Example of AI-generated hallucinated text compared with the original journal page*

AI supported this process in several ways: assisting with transcription, layout development, and improving readability, as well as contributing to visual engagement through features such as image colourisation. AI tools, for example, were used to add colour to sketches and photographs, making them feel more immediate and helping users better visualise the people and scenes described. AI was also used to assist in drafting concise, accessible labels, suited to a museum-style environment. These outputs were refined and edited to ensure they aligned with the tone, interpretation, and aims of the project. It was further used in a more analytical way, supporting observations such as comparing facial features in photographs and linking them to signatures within the journal. This helped refine my understanding of the material, for example, identifying that the same individuals appeared repeatedly rather than representing a larger group, which shaped how the content was interpreted and presented.

*Figure 13: Example of AI-assisted image analysis and comparison of facial features*

This was important to show, as whilst AI can support the process, it has no understanding of historical context or meaning. The interpretation, selection, and overall direction of the project remained entirely human, thereby it reflects a collaborative process, with AI extending what is possible without replacing human judgement. Making the use of AI visible within the design provides users some insight into usage of the tools, helping them understand the relationship between human decision making and AI support as something that enhances access rather than replaces human roles. It also introduces AI as a creative tool for those who may not have encountered it before.

AI was an essential part of this project as without it, much of the journal would have remained inaccessible to me. The AI-supported transcriptions acted as a starting point, which I then compared carefully against the original manuscript. Over time, this process improved my own ability to read the handwriting, as I began to recognise patterns and challenge myself to interpret independent of AI. In this sense, the use of AI did not replace my understanding, but supported and developed it, making it a valuable tool within the overall process.

## Implementation 

### Opening Pages and Visual Experience

I set out to create the design in Figma, with a clear vision of how the project should look and how users would navigate through the book. Rather than spending large amounts of time building the layout manually, I made use of Figma’s AI features, providing specific instructions for elements such as layout, font, and cover design. This allowed me to quickly generate a draft of key pages, using basic information about the object and an image to form an initial template. While this use of AI is not something I am formally assessing within the project, it played an important role in supporting the design process and helped bring my ideas to life in a format suited to a museum-style digital exhibition.


*Figure 14: Early Figma layout draft generated using AI-assisted design tools*

I decided to recreate the opening pages of the journal allowing all users to begin with the same experience, mirroring the physical book. Using Gemini, I generated a full image of the book’s cover that could be adapted as a digital front cover. I then recreated the first page, featuring the photograph The Wave taken from Row Quay, 10 July 1865. In addition to the archival image, I used Gemini to add colour to the photograph which required several attempts. The initial results appeared too modern and needed more specific prompts to ensure only colour was added to the existing image.

*Figure 15: Opening page featuring The Wave with archival and colourised versions*

Users are prompted to click on the image, to view both the original and colourised versions side by side. ChatGPT was used to clean up my photograph of the original page making it appear more like a clear scan, while still preserving the texture and marks of the paper. The page includes labels distinguishing between archival and AI-assisted content, a short descriptive passage introducing the scene, and a note explaining that the colourisation is interpretive rather than historically accurate. A short reflective prompt is also included to encourage users to think about who would have had access to photography at the time, introducing themes of class early in the project.

*Figure 16: Interactive comparison between archival and colourised image*

### Portrait Analysis

Several pages of the journal include portrait photographs of individuals likely connected to the journeys. These were purposefully brought them together in a gallery for easier comparison. By analysing facial features and matching signatures, supported by AI tools, I concluded that the images most likely show two individuals: J. Smith and John Laurie. This remains an informed interpretation rather than confirmed fact and is clearly stated within the project. AI was also used to clean and extend the images for digital presentation, helping them sit more clearly within the layout while maintaining their original archival quality.


*Figure 17: Portrait gallery used for comparative analysis of individuals*

## Ethical Considerations & Limitations

Throughout the project it was important for me to be transparent about the use of AI and the limits of interpretation. Content is clearly labelled as archival, edited, or AI-assisted enabling users to distinguish between original material and supported elements. AI-generated outputs, such as transcriptions, summaries, and colourised images, were reviewed and refined to ensure they remained as accurate as possible, remaining mindful of the limitations in their reliability and historical accuracy. Colourisation, for example, is used as an interpretive tool rather than a factual reconstruction and identifications made through image analysis remain informed assumptions rather than confirmed facts.

The scope of the project is also limited. Due to the volume and complexity of the journal, only selected extracts are included, with some content inevitably left out. AI tools also struggled with handwriting recognition, reinforcing the need for human judgement throughout. These limitations highlight AI’s ability to support access and interpretation but inability to replace critical understanding or contextual knowledge.
